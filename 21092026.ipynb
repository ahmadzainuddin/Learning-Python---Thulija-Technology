```python
# Problem:
# Doctor says when somebody comes to my click the system must greet them.
# Computer must greet them as "Welcome to Klink Kita"

# Solution:
# If you want computer to say this then you must write an instruction.
# In python they have a built in function called print.
# When you call this print function it can print anything you pass to this function.

# What to do:
# You have to call the print function and pass the value "Welcome to Klink Kita"

# How to do:
# How can i call the built in function ?
# How to pass the values which we want to be printed ?

# To call any python function mention the python function name followed by ()
# Pass the value "Welcome to Klink Kita" between ()
# In computer science we call the value(s) as "Literals"
# Every literal has its own data type.
# If your value is alphabet (english word / sentence),
# then the data type of the value is string literal.
# In python the String Literals are always enclosed with double quote ""
print("Welcome to Klink Kita")
# In programming we call the above line as statement (instruction to the computer)
```

# Literals

```python
print("Xofluza") # "Xofluza" is string literal
print(6) # number without decimal 6 is an integer literal
print(6 * 2.5 + 50) # number with decimal is float literal
# We are passing an equation to the print function
# What ever inside the () will get evaluated first
# We are multipling the medicine quantity with medicine price and then we add 50 as doctor consultation fees
# This is evaluated first which returns a value and that value is passed to the print function
# the print function prints it.
```

```python
# The doctor wants to print Medicine: Xofluza
# The doctor wants to print Quantity: 6
# The doctor wants to print Amount to be paid: 65.0

# Can we pass more than a single value to the print function ?
# You can pass more than a single value to the print function and the values are seperated by comma
print("Medicine:", "Xofluza")
print("Quantity:", 6)
print("Amount to be paid:", 65.0)

# Over here we are passing "Medicine:" and "Xofluza" they are called "arguments"
```

```python
# Is the clinic Open or Not ?
# For such questions usually we will say Yes / No
# In computer for this type of question we answer it in Boolean
# Yes => True / No => False
print(True) # meaning yes it is open
# True is also an english word if alphabet it must be enclosed with double quote ""
# However True / False are keywords of Python. Meaning python know them. They are special words.
# Keywords are not enclosed with double quote
print("False") # python will treat that as string over here that is ok
# However when you want to make a decision True is totally different from "True"
```

```python
# I am bought few fruits apple, orange, mango, banana and grapes
# how do i print them
# I want to place them inside a box [] this is called list in python
print(["apple", "orange", "mango", "banana", "grapes"])
# We group them already and they are not individual items
# Now let us pass the box to the print function
```

```python
# Can we pass a value (literal) to python and ask python to identify the literal is what data type ?
# There is a built in function called type which takes literal as argument
# The type function will return the data type of that literal
# Even though type function returns data type we still have to call print function to print data type
print(type("Xofluza"))
print(type(6))
print(type(2.5))
print(type(True))
```

image.png

```python
print(type(["apple", "orange", "mango", "banana", "grapes"])) # [] is list
# list is modifiable (add, remove, insert, update, delete)
# list allows duplicate
print(type(("apple", "orange", "mango", "banana", "grapes"))) # () is tuple
# tuple is read only list (not modifiable)
# tuple allows duplicate
print(type({"apple", "orange", "mango", "banana", "grapes"})) # {} is set
# set does not allow duplicate
```

# Variables

image.png

```python
medicine = "Xofluza"
# Here medicine is a variable
# xofluza is a literal
# As a human we go to school and we aleady learn left side is a variable
# and right side may variable or literal
# As a human we know how to identify them
# However python do not know which is variable and which is literal

# To differentiate string literals from the variables
# the string literals are enclosed with double quote
quantity = 6 # quantity is variable 6 is literal
price = 2.5 # price is variable 2.5 is literal
is_open = True # is_open is variable True is literal
```

```python
print(medicine)
print(quantity)
print(price)
print(is_open)
```

```python
# The variable is not holding the value
# the variable is holding the address location
# Can I see the address location ?
print(medicine, id(medicine), type(medicine))
print(quantity, id(quantity), type(quantity))
print(price, id(price), type(price))
print(is_open, id(is_open), type(is_open))
```

```python
x = 10
print(x, id(x), type(x))
x = 15
print(x, id(x), type(x))
y = 15 # technically it suppose to create another object 15
print(y, id(y), type(y))
# however python does memory scanning and if the object happens to be in memory
# python will not create that object again rather it will just reuse them
x = "Television"
print(x, id(x), type(x))
print(y, id(y), type(y))
```
