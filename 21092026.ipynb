```python
# find whether the given fruit is inside the list
givenfruit = "orange"
fruits = ["apple", "orange", "mango", "banana", "grapes"]
for fruit in fruits:
  if fruit == givenfruit:
    print(givenfruit, "is found")
    break

# For efficiency we add a break keyword inside the loop so that we come out of the loop
# the moment we found the fruit. No point iterating the remaining fruits
```

## While Loop

- Do you have a list ? - No
- Do you know how many times we have to execute block of code ? - No

You have to use While Loop

### Example

- We want to get integer input
- User key in invalid input
- System throws error
- System ask the use to key valid input
- This will be repeated (we do not know how many times)
- Because it depends on the users behaviour

```python
# Let us learn how to get user input ?
# In python we have a built in function input
quantity = input("Enter the quantity:")
print(quantity)
print(type(quantity))

# Regardless of what the value keyed in the input function will always return string data type
# If you want integer you must typecast it
```

```python
quantity = input("Key in quantity:")
# The error occurs when you try to type cast the user input
# However if the user key in integer the type cast is successful
# The code got no issue. The issue is because of user input.
# This type of error we call it as runtime exception
quantity = int(quantity)

# How do we handle runtime exception ?
# In python we have try except else finally block
```

```python
try:
  # the code that can cause a runtime error should be placed
  # inside the try block
  quantity = input("Key in quantity:")
  quantity = int(quantity)
except:
  # When the code in the try block throws error. what we should do ?
  # that code should be placed inside the except block
  print("Quantity must be Integer")
else:
  # When the code in the try block executed no error occur. what we should do ?
  # that code should be placed inside the else block
  print("Valid Quantity")
finally:
  # this code always gets executed regardless of whether the an error occur
  # or no error occur
  print("Thank you")

print(quantity)
```

```python
# Technically we should not accept abc rather we should force the user to key in
# integer value. This is where we going to use while loop do it until the user key
# in valid input.
# Here we are creating a while loop that will keep repeating indefinetely

# isInvalidInput = True
# while isInvalidInput == True
while True:
  try:
    quantity = input("Key in quantity:")
    quantity = int(quantity)
  except:
    print("Please key in integer value")
  else:
    print("Valid Quantity")
    # isInvalidInput = False
    break # here we come out of the loop using break keyword
  finally:
    print("Thank you")

print(quantity)
```
